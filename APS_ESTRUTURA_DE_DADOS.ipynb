```python
from google.colab import files

import pandas as pd

def carregar_dados():

    arquivo = files.upload()

    nome = list(arquivo.keys())[0]

    df = pd.read_excel(nome)

    coluna = next(c for c in df.columns if str(c).strip().lower() == "areamunkm")

    return df, coluna

df, coluna = carregar_dados()

print("Visualização parcial da planilha:")
display(df.head())
```

INVERSAMENTE ORDENADA

```python
dados = df.sort_values(
    by=coluna,
    ascending=False
).to_dict("records")

print("Planilha após ordenação decrescente (PIOR CASO - TROCA DE TODOS OS ELEMENTOS):")
display(pd.DataFrame(dados).head())
```

ALEATÓRIA

```python
import random

dados_aleatorios = df.to_dict("records")

random.shuffle(dados_aleatorios)

print("Planilha após embaralhamento aleatório:")
display(pd.DataFrame(dados_aleatorios).head())
```

BUBBLE SORT

```python
import time

resultados = [
    ["Algoritmo", "Inversamente ordenada", "Aleatória"],
]

def bubble_sort(dados):

    dados = dados.copy()

    for i in range(len(dados) - 1):

        for j in range(len(dados) - 1 - i):

            if dados[j][coluna] > dados[j + 1][coluna]:

                dados[j], dados[j + 1] = dados[j + 1], dados[j]

    return dados


inicio = time.perf_counter()

resultado_inverso = bubble_sort(dados)

fim = time.perf_counter()

tempo_inverso = fim - inicio


inicio = time.perf_counter()

resultado_aleatorio = bubble_sort(dados_aleatorios)

fim = time.perf_counter()

tempo_aleatorio = fim - inicio


resultados.append([
    "Bubble Sort",
    tempo_inverso,
    tempo_aleatorio
])


diferenca = tempo_inverso - tempo_aleatorio


print("BUBBLE SORT")
print()
print(f"Simulação 1 - Inversamente ordenada: {tempo_inverso:.2e} segundos")
print(f"Simulação 2 - Aleatória:              {tempo_aleatorio:.2e} segundos")
print()
print(f"Diferença entre os tempos: {abs(diferenca):.2e} segundos")


if diferenca > 0:

    print(
        f"O algoritmo gastou {diferenca:.2e} segundos a mais "
        f"na situação inversamente ordenada."
    )

elif diferenca < 0:

    print(
        f"O algoritmo gastou {abs(diferenca):.2e} segundos a mais "
        f"na situação aleatória."
    )

else:

    print("O algoritmo apresentou o mesmo tempo nas duas situações.")
```

SELECTION SORT

```python
import time

def selection_sort(dados):

    dados = dados.copy()

    for i in range(len(dados) - 1):

        menor = i

        for j in range(i + 1, len(dados)):

            if dados[j][coluna] < dados[menor][coluna]:
                menor = j

        dados[i], dados[menor] = dados[menor], dados[i]

    return dados


inicio = time.perf_counter()

resultado_inverso = selection_sort(dados)

fim = time.perf_counter()

tempo_inverso = fim - inicio


inicio = time.perf_counter()

resultado_aleatorio = selection_sort(dados_aleatorios)

fim = time.perf_counter()

tempo_aleatorio = fim - inicio


resultados.append([
    "Selection Sort",
    tempo_inverso,
    tempo_aleatorio
])


diferenca = tempo_inverso - tempo_aleatorio


print("SELECTION SORT")
print()
print(f"Simulação 1 - Inversamente ordenada: {tempo_inverso:.2e} segundos")
print(f"Simulação 2 - Aleatória:              {tempo_aleatorio:.2e} segundos")
print()
print(f"Diferença entre os tempos: {abs(diferenca):.2e} segundos")


if diferenca > 0:

    print(
        f"O algoritmo gastou {diferenca:.2e} segundos a mais "
        f"na situação inversamente ordenada."
    )

elif diferenca < 0:

    print(
        f"O algoritmo gastou {abs(diferenca):.2e} segundos a mais "
        f"na situação aleatória."
    )

else:

    print("O algoritmo apresentou o mesmo tempo nas duas situações.")
```

INSERTION SORT

```python
import time

def insertion_sort(dados):

    dados = dados.copy()

    for i in range(1, len(dados)):

        atual = dados[i]

        j = i - 1

        while j >= 0 and dados[j][coluna] > atual[coluna]:

            dados[j + 1] = dados[j]

            j -= 1

        dados[j + 1] = atual

    return dados


inicio = time.perf_counter()

resultado_inverso = insertion_sort(dados)

fim = time.perf_counter()

tempo_inverso = fim - inicio


inicio = time.perf_counter()

resultado_aleatorio = insertion_sort(dados_aleatorios)

fim = time.perf_counter()

tempo_aleatorio = fim - inicio


resultados.append([
    "Insertion Sort",
    tempo_inverso,
    tempo_aleatorio
])


diferenca = tempo_inverso - tempo_aleatorio


print("INSERTION SORT")
print()
print(f"Simulação 1 - Inversamente ordenada: {tempo_inverso:.2e} segundos")
print(f"Simulação 2 - Aleatória:              {tempo_aleatorio:.2e} segundos")
print()
print(f"Diferença entre os tempos: {abs(diferenca):.2e} segundos")


if diferenca > 0:

    print(
        f"O algoritmo gastou {diferenca:.2e} segundos a mais "
        f"na situação inversamente ordenada."
    )

elif diferenca < 0:

    print(
        f"O algoritmo gastou {abs(diferenca):.2e} segundos a mais "
        f"na situação aleatória."
    )

else:

    print("O algoritmo apresentou o mesmo tempo nas duas situações.")
```

MERGE SORT

```python
import time

def merge_sort(dados):

    if len(dados) <= 1:
        return dados.copy()

    meio = len(dados) // 2

    esquerda = merge_sort(dados[:meio])
    direita = merge_sort(dados[meio:])

    resultado = []

    i = 0
    j = 0

    while i < len(esquerda) and j < len(direita):

        if esquerda[i][coluna] <= direita[j][coluna]:

            resultado.append(esquerda[i])

            i += 1

        else:

            resultado.append(direita[j])

            j += 1

    resultado.extend(esquerda[i:])
    resultado.extend(direita[j:])

    return resultado


inicio = time.perf_counter()

resultado_inverso = merge_sort(dados)

fim = time.perf_counter()

tempo_inverso = fim - inicio


inicio = time.perf_counter()

resultado_aleatorio = merge_sort(dados_aleatorios)

fim = time.perf_counter()

tempo_aleatorio = fim - inicio


resultados.append([
    "Merge Sort",
    tempo_inverso,
    tempo_aleatorio
])


diferenca = tempo_inverso - tempo_aleatorio


print("MERGE SORT")
print()
print(f"Simulação 1 - Inversamente ordenada: {tempo_inverso:.2e} segundos")
print(f"Simulação 2 - Aleatória:              {tempo_aleatorio:.2e} segundos")
print()
print(f"Diferença entre os tempos: {abs(diferenca):.2e} segundos")


if diferenca > 0:

    print(
        f"O algoritmo gastou {diferenca:.2e} segundos a mais "
        f"na situação inversamente ordenada."
    )

elif diferenca < 0:

    print(
        f"O algoritmo gastou {abs(diferenca):.2e} segundos a mais "
        f"na situação aleatória."
    )

else:

    print("O algoritmo apresentou o mesmo tempo nas duas situações.")
```

QUICK SORT

```python
import time

def quick_sort(dados):

    if len(dados) <= 1:
        return dados.copy()

    pivo = dados[len(dados) // 2][coluna]

    menores = []
    iguais = []
    maiores = []

    for registro in dados:

        if registro[coluna] < pivo:

            menores.append(registro)

        elif registro[coluna] > pivo:

            maiores.append(registro)

        else:

            iguais.append(registro)

    return (
        quick_sort(menores)
        + iguais
        + quick_sort(maiores)
    )


inicio = time.perf_counter()

resultado_inverso = quick_sort(dados)

fim = time.perf_counter()

tempo_inverso = fim - inicio


inicio = time.perf_counter()

resultado_aleatorio = quick_sort(dados_aleatorios)

fim = time.perf_counter()

tempo_aleatorio = fim - inicio


resultados.append([
    "Quick Sort",
    tempo_inverso,
    tempo_aleatorio
])


diferenca = tempo_inverso - tempo_aleatorio


print("QUICK SORT")
print()
print(f"Simulação 1 - Inversamente ordenada: {tempo_inverso:.2e} segundos")
print(f"Simulação 2 - Aleatória:              {tempo_aleatorio:.2e} segundos")
print()
print(f"Diferença entre os tempos: {abs(diferenca):.2e} segundos")


if diferenca > 0:

    print(
        f"O algoritmo gastou {diferenca:.2e} segundos a mais "
        f"na situação inversamente ordenada."
    )

elif diferenca < 0:

    print(
        f"O algoritmo gastou {abs(diferenca):.2e} segundos a mais "
        f"na situação aleatória."
    )

else:

    print("O algoritmo apresentou o mesmo tempo nas duas situações.")
```

GRÁFICO DE DADOS INVERSAMENTE ORDENADOS (ORDENADOS DE FORMA DECRESCENTE PARA ORDENAÇÃO CRESCENTE)

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
from matplotlib.ticker import FuncFormatter, AutoMinorLocator

tabela_resultados = pd.DataFrame(
    resultados[1:],
    columns=resultados[0]
)

algoritmos = tabela_resultados["Algoritmo"].tolist()

x = np.arange(len(algoritmos))

cores_base = [
    "#1565C0",
    "#EF6C00",
    "#2E7D32",
    "#C62828",
    "#6A1B9A",
    "#00838F",
    "#AD1457",
    "#6D4C41",
    "#283593",
    "#558B2F"
]

marcadores_base = [
    "o",
    "s",
    "^",
    "D",
    "P",
    "X",
    "*",
    "v",
    "<",
    ">"
]

cores = [
    cores_base[i % len(cores_base)]
    for i in range(len(algoritmos))
]

marcadores = [
    marcadores_base[i % len(marcadores_base)]
    for i in range(len(algoritmos))
]


def criar_grafico(coluna_tempo, titulo_situacao):

    tempos = tabela_resultados[coluna_tempo].tolist()

    fig, ax = plt.subplots(figsize=(15, 9))

    ax.plot(
        x,
        tempos,
        color="#263238",
        linewidth=3.5,
        linestyle="-",
        alpha=0.9,
        zorder=2
    )

    for i, (algoritmo, tempo) in enumerate(zip(algoritmos, tempos)):

        ax.scatter(
            i,
            tempo,
            s=180,
            color=cores[i],
            marker=marcadores[i],
            edgecolor="white",
            linewidth=2.5,
            zorder=5,
            label=algoritmo
        )

        ax.annotate(
            f"{tempo:.2e} s",
            (i, tempo),
            xytext=(0, 18),
            textcoords="offset points",
            ha="center",
            va="bottom",
            fontsize=11,
            fontweight="bold",
            color=cores[i],
            bbox=dict(
                boxstyle="round,pad=0.35",
                facecolor="white",
                edgecolor=cores[i],
                linewidth=1.5
            )
        )

    ax.set_xticks(x)

    ax.set_xticklabels(
        algoritmos,
        fontsize=12,
        fontweight="bold"
    )

    ax.set_title(
        "Comparação do Tempo de Execução dos Algoritmos",
        fontsize=21,
        fontweight="bold",
        pad=25
    )

    ax.text(
        0.5,
        1.015,
        titulo_situacao,
        transform=ax.transAxes,
        ha="center",
        fontsize=14,
        color="#555555"
    )

    ax.set_xlabel(
        "Algoritmo de ordenação",
        fontsize=14,
        fontweight="bold",
        labelpad=12
    )

    ax.set_ylabel(
        "Tempo de execução (segundos)",
        fontsize=14,
        fontweight="bold",
        labelpad=12
    )

    ax.yaxis.set_major_formatter(
        FuncFormatter(lambda valor, pos: f"{valor:.1e}")
    )

    ax.yaxis.set_minor_locator(
        AutoMinorLocator(5)
    )

    ax.grid(
        which="major",
        axis="both",
        linestyle="-",
        linewidth=0.8,
        alpha=0.30
    )

    ax.grid(
        which="minor",
        axis="both",
        linestyle=":",
        linewidth=0.6,
        alpha=0.20
    )

    ax.set_axisbelow(True)

    ax.set_facecolor("#F7F9FC")
    fig.patch.set_facecolor("white")

    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)

    ax.legend(
        title="Algoritmos",
        title_fontsize=12,
        fontsize=11,
        loc="upper left",
        frameon=True,
        fancybox=True,
        shadow=True
    )

    ax.set_xlim(
        -0.5,
        len(algoritmos) - 0.5
    )

    plt.tight_layout()

    plt.show()
```

```python
criar_grafico(
    "Inversamente ordenada",
    "Simulação 1 — Dados inversamente ordenados"
)
```
